```python
import pandas as pd
import torch
import os

df = pd.read_excel("encoder_instance_features.xlsx")

feature_cols = [c for c in df.columns if c.startswith("feat_")]


for pid, group in df.groupby("patient_id"):
    feats = group[feature_cols].values  # shape [N, D]
    feats = torch.tensor(feats, dtype=torch.float32)

    torch.save(
        {"patient_id": pid, "feats": feats},
        f"encoder_features/{pid}.pt"
    )

print("DONE — saved per-patient features in features_pt/")
```

```python
import os
import torch
import numpy as np

FEATURE_DIR = "encoder_features"

patient_counts = []     # list of N values
feature_dims = set()    # set of possible D values
all_values_sample = []  # small sample of values for global stats

print("Inspecting:", FEATURE_DIR)
print("-" * 60)

for fname in os.listdir(FEATURE_DIR):
    if not fname.endswith(".pt"):
        continue

    path = os.path.join(FEATURE_DIR, fname)
    data = torch.load(path)

    pid = data["patient_id"]
    feats = data["feats"]   # [N, D]

    N, D = feats.shape
    patient_counts.append(N)
    feature_dims.add(D)

    # sample a few values for global range
    sample_vals = feats.flatten()[:100].cpu().numpy()
    all_values_sample.append(sample_vals)

    print(f"Patient {pid}: N={N}, D={D}")

print("\n" + "="*60)
print("SUMMARY")
print("="*60)

patient_counts = np.array(patient_counts)
all_values_sample = np.concatenate(all_values_sample)

print(f"Total patients: {len(patient_counts)}")
print(f"Feature dims found: {feature_dims}")
print()

print(f"Min images per patient: {patient_counts.min()}")
print(f"Max images per patient: {patient_counts.max()}")
print(f"Mean images per patient: {patient_counts.mean():.2f}")
print(f"Median images per patient: {np.median(patient_counts)}")
print()

print(f"Feature values: min={all_values_sample.min():.6f}, max={all_values_sample.max():.6f}")
print(f"Feature mean: {all_values_sample.mean():.6f}")
print(f"Feature std:  {all_values_sample.std():.6f}")
print()
```

```python
import torch
import numpy as np

pid = 107  # change this to the patient ID you want to inspect

path = f"encoder_features/{pid}.pt"
data = torch.load(path)

print("=== PATIENT INFO ===")
print("Patient ID:", data["patient_id"])

feats = data["feats"]  # [N, D]
print("Feats type:", type(feats))
print("Feats shape:", feats.shape)  # (N, D)

N, D = feats.shape
print("Images (N):", N)
print("Feature dim (D):", D)

# convert to numpy for stats
arr = feats.cpu().numpy()

print("\n=== FEATURE STATISTICS (ALL IMAGES) ===")
for i in range(N):
    v = arr[i]
    print(f"Image #{i+1:03d}: "
          f"min={v.min():.6f}, "
          f"max={v.max():.6f}, "
          f"mean={v.mean():.6f}, "
          f"std={v.std():.6f}")
```

```python
import os
import torch
import pandas as pd
import numpy as np

feat_dir = "./encoder_features"
out_dir = "./outputs"
n_folds = 5   # change if needed

def fold_stats(fold):
    val_csv = os.path.join(out_dir, f"valid_fold{fold}.csv")
    df = pd.read_csv(val_csv)
    ids = df["id"].astype(str).tolist()

    counts = []
    for pid in ids:
        path = os.path.join(feat_dir, f"{pid}.pt")
        if not os.path.exists(path):
            continue
        data = torch.load(path, weights_only=True)
        counts.append(data["feats"].shape[0])

    counts = np.array(counts)

    print(f"\n===== Fold {fold} Validation Statistics =====")
    print(f"Patients        : {len(counts)}")
    print(f"Min images      : {counts.min()}")
    print(f"Max images      : {counts.max()}")
    print(f"Mean images     : {counts.mean():.2f}")
    print(f"Median images   : {np.median(counts)}")
    print(f"Std images      : {counts.std():.2f}")

    bins = [0, 5, 10, 20, 32, 50, 100, 200, 500, 1000]
    hist = np.histogram(counts, bins=bins)

    print("Image count distribution:")
    for i in range(len(bins) - 1):
        print(f"  {bins[i]:>4} – {bins[i+1]:<4}: {hist[0][i]}")

# run all folds
for fold in range(1, n_folds + 1):
    fold_stats(fold)
```
